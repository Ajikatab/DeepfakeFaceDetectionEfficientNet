# 🧪 Notebook Simulasi Pengujian Komparatif 12 Skenario (Tanpa GUI)
Notebook ini mengeksekusi pengujian foto pada folder **`FOTO SIMULASI`** secara komparatif melintasi **seluruh 12 Skenario Model Deep Learning** (EfficientNet-B0, B3, B7 x RMSprop, Adam, SGD, AdamW).

```python
import os
import sys
import time
import numpy as np
import tensorflow as tf
from PIL import Image
import pandas as pd

# Import fungsi resmi dari gui_tester (100% Arsitektur Presisi)
sys.path.append(os.path.abspath('.'))
from gui_tester import MODEL_CONFIGS, build_exact_training_model, load_exact_weights

print("TensorFlow Version:", tf.__version__)
print("Simulasi Pengujian 12 Skenario Siap Eksekusi!")
```

```python
# 1. Persiapan Data Gambar dari Folder 'FOTO SIMULASI'
SIMULATION_DIR = os.path.abspath('FOTO SIMULASI')
image_files = sorted([f for f in os.listdir(SIMULATION_DIR) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])

ground_truths = []
for fname in image_files:
    if fname.startswith('ga'):
        gt = "ASLI (REAL)"
    elif fname.startswith(('kb', 'ekb')):
        gt = "AI (FAKE)"
    else:
        gt = "UNKNOWN"
    ground_truths.append(gt)

print(f"Total Gambar Simulasi : {len(image_files)} Citra")
print(f"-> Foto ASLI (REAL)    : {ground_truths.count('ASLI (REAL)')} Citra")
print(f"-> Foto AI (FAKE)      : {ground_truths.count('AI (FAKE)')} Citra")
```

```python
# 2. Eksekusi Pengujian Komparatif Seluruh 12 Skenario Model
summary_results = []
detail_rows = {fname: {'Nama File': fname, 'Kategori Asli': ground_truths[i]} for i, fname in enumerate(image_files)}

scenario_num = 1
for variant_name, cfg in MODEL_CONFIGS.items():
    img_size = cfg['img_size']
    
    # Pre-process image batch untuk varian saat ini
    batch_arrs = []
    for fname in image_files:
        fpath = os.path.join(SIMULATION_DIR, fname)
        img = Image.open(fpath).convert('RGB')
        img_resized = img.resize(img_size, getattr(Image, 'Resampling', Image).NEAREST)
        img_arr = tf.keras.preprocessing.image.img_to_array(img_resized)
        batch_arrs.append(img_arr)
    images_batch = np.array(batch_arrs)
    
    for opt_name, pattern in cfg['patterns'].items():
        best_fold = cfg['best_folds'][opt_name]
        weight_path = os.path.join(cfg['folder'], f"{pattern}fold{best_fold}.weights.h5")
        
        t0 = time.time()
        model, _ = build_exact_training_model(variant_name)
        load_exact_weights(model, weight_path)
        
        preds_batch = model(images_batch, training=False).numpy()
        elapsed_t = time.time() - t0
        
        correct_cnt = 0
        ai_correct_cnt = 0
        asli_correct_cnt = 0
        
        for i, fname in enumerate(image_files):
            preds = preds_batch[i]
            pred_idx = int(np.argmax(preds))
            pred_label = "ASLI (REAL)" if pred_idx == 1 else "AI (FAKE)"
            gt = ground_truths[i]
            
            is_correct = (pred_label == gt)
            if is_correct:
                correct_cnt += 1
                if gt == "AI (FAKE)":
                    ai_correct_cnt += 1
                else:
                    asli_correct_cnt += 1
                    
            detail_rows[fname][f"{variant_name} + {opt_name}"] = f"{pred_label} ({max(preds)*100:.1f}%)"

        acc_pct = (correct_cnt / len(image_files)) * 100
        
        summary_results.append({
            'No': scenario_num,
            'Arsitektur Model': variant_name,
            'Optimizer': opt_name,
            'Fold Terpakai': f"Fold {best_fold}",
            'Total Benar': f"{correct_cnt} / {len(image_files)}",
            'Akurasi Simulasi (%)': f"{acc_pct:.2f}%":
            'Deteksi ASLI Tepat': f"{asli_correct_cnt} / 8",
            'Deteksi AI Tepat': f"{ai_correct_cnt} / 5",
            'Waktu Pros (s)': f"{elapsed_t:.2f} s"
        })
        
        print(f"Skenario {scenario_num:2d}: {variant_name:15s} | {opt_name:20s} (Fold {best_fold}) -> Akurasi: {acc_pct:6.2f}% ({correct_cnt}/{len(image_files)})")
        scenario_num += 1
```

```python
# 3. Tampilkan Tabel Ringkasan Komparatif 12 Skenario
df_summary = pd.DataFrame(summary_results)
print("=== RINGKASAN AKURASI PENGUJIAN SIMULASI PER SKENARIO ===")
df_summary
```
